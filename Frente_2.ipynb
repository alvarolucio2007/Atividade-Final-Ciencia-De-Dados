```python
colunas_datas_pedidos = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for coluna in colunas_datas_pedidos:
    df_orders[coluna] = pd.to_datetime(df_orders[coluna], errors="coerce")

# Conversão da dtta em itens
df_order_items["shipping_limit_date"] = pd.to_datetime(
    df_order_items["shipping_limit_date"], errors="coerce"
)

# Converão ds datas das avaliações
df_order_reviews["review_creation_date"] = pd.to_datetime(
    df_order_reviews["review_creation_date"], errors="coerce"
)

df_order_reviews["review_answer_timestamp"] = pd.to_datetime(
    df_order_reviews["review_answer_timestamp"], errors="coerce"
)

print("Datas dos pedidos:")
print(df_orders[colunas_datas_pedidos].dtypes)

print("\n")
print("Datas dos itens e avaliações convertidas!")
```

```python
df_products["product_category_name"] = df_products["product_category_name"].fillna("sem_categoria")

print("Categorias ausentes:", df_products["product_category_name"].isnull().sum())

print("Categorias diferentes:", df_products["product_category_name"].nunique())
```

```python
print("Pediddos repetidos ns avaliações:")
print(df_order_reviews["order_id"].duplicated().sum())
```

```python
avaliacoes_por_pedido = df_order_reviews.groupby("order_id").agg(
    quantidade_avaliacoes=("review_id", "count"),
    notas_diferentes=("review_score", "nunique")
)

pedidos_repetidos = avaliacoes_por_pedido[
    avaliacoes_por_pedido["quantidade_avaliacoes"] > 1
]

print("Pedidos com mais de uma avaliação:", len(pedidos_repetidos))

print("\n")
print("Quantidade de notas diferentes nesses pedidos:")
print(pedidos_repetidos["notas_diferentes"].value_counts())

print("\n")
print("\nExemplos:")
display(pedidos_repetidos.head(10))
```

```python
df_reviews = df_order_reviews.groupby("order_id").agg(
    review_score=("review_score", "mean")
).reset_index()

print("Total de pedidos com avaliações:", len(df_reviews))
print("Pedidos duplicados:", df_reviews["order_id"].duplicated().sum())

display(df_reviews.head())
```

```python
df_analise = df_order_items.merge(
    df_products[["product_id", "product_category_name"]],
    on="product_id",
    how="left"
)

print("Total de itens:", len(df_order_items))
print("Total após cruzamento:", len(df_analise))

print("Primeiras linhas:")
display(df_analise.head())
```

```python
# Remover colunas que já existem na base
colunas_orders = [
    "order_status",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

df_analise = df_analise.drop(
    columns=[
        coluna for coluna in colunas_orders
        if coluna in df_analise.columns
    ]
)

# Adicionar informações dos pedidos
df_analise = df_analise.merge(
    df_orders[
        [
            "order_id",
            "order_status",
            "order_delivered_customer_date",
            "order_estimated_delivery_date"
        ]
    ],
    on="order_id",
    how="left"
)

# Adicionar avaliações consolidadas
df_analise = df_analise.merge(
    df_reviews,
    on="order_id",
    how="left"
)

print("Total de registros:", len(df_analise))

print("\nColunas da base de análise:")
print(df_analise.columns.tolist())

print("\nPrimeiras linhas:")
display(df_analise.head())
```

```python
print("Valores ausentes na base de análise:")
print(df_analise[
    [
        "product_category_name",
        "order_status",
        "order_delivered_customer_date",
        "order_estimated_delivery_date",
        "price",
        "freight_value",
        "review_score"
    ]
].isnull().sum())

print("\nStatus dos pedidos:")
print(df_analise["order_status"].value_counts())
```

```python
df_analise["pedido_atrasado"] = (
    (df_analise["order_status"] == "delivered") &
    (df_analise["order_delivered_customer_date"] >
     df_analise["order_estimated_delivery_date"])
).astype(int)

print("Quantidade de itens com pedido atrasado:")
print(df_analise["pedido_atrasado"].sum())

print("\nDistribuição:")
print(df_analise["pedido_atrasado"].value_counts())
```

```python
df_analise["avaliacao_baixa"] = (
    df_analise["review_score"] <= 2
).astype("Int64")

# Preservar avaliações ausentes
df_analise.loc[
    df_analise["review_score"].isnull(),
    "avaliacao_baixa"
] = pd.NA

print("Quantidade de avaliações baixas:")
print(df_analise["avaliacao_baixa"].sum())

print("\nDistribuição:")
print(df_analise["avaliacao_baixa"].value_counts(dropna=False))
```

```python
df_analise.loc[
    (df_analise["order_status"] != "delivered") |
    (df_analise["order_delivered_customer_date"].isnull()),
    "pedido_atrasado"
] = pd.NA

df_analise["pedido_atrasado"] = df_analise["pedido_atrasado"].astype("Int64")

print("Distribuição dos atrasos:")
print(df_analise["pedido_atrasado"].value_counts(dropna=False))
```

```python
print(df_analise.columns.tolist())
```

```python
# 1. Criar indicador de pedidos atrasados

df_analise["pedido_atrasado"] = np.where(
    df_analise["order_delivered_customer_date"].notna() &
    df_analise["order_estimated_delivery_date"].notna(),

    (
        df_analise["order_delivered_customer_date"] >
        df_analise["order_estimated_delivery_date"]
    ).astype(int),

    np.nan
)

# 2. Criar indicador de avaliações baixas (notas 1 e 2)

df_analise["avaliacao_baixa"] = np.where(
    df_analise["review_score"].notna(),

    (df_analise["review_score"] <= 2).astype(int),

    np.nan
)

# 3. Agrupar os indicadores por categoria

df_categorias = df_analise.groupby(
    "product_category_name",
    observed=True
).agg(
    quantidade_itens=("order_item_id", "count"),
    quantidade_pedidos=("order_id", "nunique"),

    itens_com_atraso=("pedido_atrasado", "count"),
    percentual_atrasos=("pedido_atrasado", "mean"),

    itens_com_avaliacao=("avaliacao_baixa", "count"),
    percentual_avaliacoes_baixas=("avaliacao_baixa", "mean"),

    percentual_frete_medio=("frete_percentual_preco", "mean")
).reset_index()

# 4. Converter os indicadores para porcentagem

df_categorias["percentual_atrasos"] *= 100

df_categorias["percentual_avaliacoes_baixas"] *= 100

# 5. Exibir resultados

print("Indicadores por categoria:")
display(df_categorias.head(10))

print("\nDimensões da tabela:")
print(df_categorias.shape)

print("\nValores ausentes:")
print(df_categorias.isnull().sum())
```

```python
# Visualizar os indicadores consolidados por categoria

print("Quantidade de categorias:", df_categorias.shape[0])

print("\nCategorias com maiores percentuais de atraso:")
display(
    df_categorias.sort_values(
        "percentual_atrasos",
        ascending=False
    ).head(10)
)

print("\nCategorias com maiores percentuais de avaliações baixas:")
display(
    df_categorias.sort_values(
        "percentual_avaliacoes_baixas",
        ascending=False
    ).head(10)
)

print("\nCategorias com maiores percentuais de frete:")
display(
    df_categorias.sort_values(
        "percentual_frete_medio",
        ascending=False
    ).head(10)
)
```

```python
# Criar uma pontuação de prioridade para cada categoria

df_categorias["volume_normalizado"] = (
    df_categorias["quantidade_itens"] /
    df_categorias["quantidade_itens"].max()
) * 100

df_categorias["pontuacao_prioridade"] = (
    df_categorias["volume_normalizado"] * 0.25 +
    df_categorias["percentual_atrasos"] * 0.25 +
    df_categorias["percentual_avaliacoes_baixas"] * 0.25 +
    df_categorias["percentual_frete_medio"].clip(upper=100) * 0.25
)

# Ordenar categorias da maior para a menor prioridade

df_categorias = df_categorias.sort_values(
    by="pontuacao_prioridade",
    ascending=False
).reset_index(drop=True)

# Exibir as 15 categorias prioritárias

print("Categorias com maior prioridade de revisão:")

display(
    df_categorias[
        [
            "product_category_name",
            "quantidade_itens",
            "percentual_atrasos",
            "percentual_avaliacoes_baixas",
            "percentual_frete_medio",
            "pontuacao_prioridade"
        ]
    ].head(15)
)
```

```python
# Classificar as categorias conforme o volume de itens

df_categorias["classificacao_volume"] = np.where(
    df_categorias["quantidade_itens"] < 100,
    "Baixo volume",
    np.where(
        df_categorias["quantidade_itens"] < 1000,
        "Médio volume",
        "Alto volume"
    )
)

# Exibir a classificação das categorias prioritárias

display(
    df_categorias[
        [
            "product_category_name",
            "quantidade_itens",
            "classificacao_volume",
            "pontuacao_prioridade"
        ]
    ].head(15)
)
```

```python
# Selecionar variáveis numéricas da base analítica

colunas_numericas = df_analise.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

# Calcular a matriz de correlação

matriz_correlacao = df_analise[
    colunas_numericas
].corr()

# Exibir a matriz de correlação

display(matriz_correlacao.round(2))
```

```python
# Identificar pares de variáveis com alta correlação

correlacao_absoluta = matriz_correlacao.abs()

# Manter apenas a parte superior da matriz para evitar pares repetidos

mascara_superior = np.triu(
    np.ones(correlacao_absoluta.shape),
    k=1
).astype(bool)

correlacoes_pares = (
    correlacao_absoluta.where(mascara_superior)
    .stack()
    .reset_index()
)

correlacoes_pares.columns = [
    "atributo_1",
    "atributo_2",
    "correlacao"
]

# Filtrar correlações fortes

correlacoes_fortes = correlacoes_pares[
    correlacoes_pares["correlacao"] >= 0.80
].sort_values(
    by="correlacao",
    ascending=False
)

print("Pares de atributos com correlação igual ou superior a 0.80:")

display(correlacoes_fortes)
```

```python
# Calcular a variância das variáveis numéricas

variancias = df_analise[colunas_numericas].var()

# Identificar atributos com variância inferior a 0.01

atributos_baixa_variancia = variancias[
    variancias < 0.01
].sort_values()

print("Quantidade de atributos com baixa variância:")
print(len(atributos_baixa_variancia))

print("\nAtributos identificados:")
display(atributos_baixa_variancia.to_frame(
    name="variancia"
))
```

```python
# Identificar as colunas geradas pelo One-Hot Encoding

colunas_one_hot = [
    coluna for coluna in df_analise.columns
    if coluna.startswith("categoria_")
]

# Calcular a frequência de cada categoria

frequencia_categorias = df_analise[colunas_one_hot].sum().sort_values()

# Exibir as 15 categorias menos frequentes

print("15 categorias com menor frequência:")

display(
    frequencia_categorias.head(15).to_frame(
        name="quantidade_registros"
    )
)

# Exibir quantidade total de colunas One-Hot

print("\nTotal de colunas One-Hot:", len(colunas_one_hot))
```

```python
# Padronização Z-Score

colunas_numericas = [
    "price",
    "freight_value",
    "frete_percentual_preco",
    "valor_total_item"
]

for coluna in colunas_numericas:

    media = df_analise[coluna].mean()
    desvio = df_analise[coluna].std()

    if desvio != 0:
        df_analise[coluna + "_zscore"] = (
            (df_analise[coluna] - media) / desvio
        )
    else:
        df_analise[coluna + "_zscore"] = 0

print("Colunas Z-Score criadas com sucesso!")

display(
    df_analise[
        [coluna + "_zscore" for coluna in colunas_numericas]
    ].head()
)
```

```python

# Normalizar os indicadores por categoria

colunas_prioridade = [
    "quantidade_itens",
    "percentual_atrasos",
    "percentual_avaliacoes_baixas"
]

for coluna in colunas_prioridade:
    minimo = df_categorias[coluna].min()
    maximo = df_categorias[coluna].max()

    df_categorias[coluna + "_normalizado"] = (
        (df_categorias[coluna] - minimo) / (maximo - minimo)
    )

# Criar pontuação de prioridade
df_categorias["pontuacao_prioridade"] = (
    df_categorias["quantidade_itens_normalizado"] +
    df_categorias["percentual_atrasos_normalizado"] +
    df_categorias["percentual_avaliacoes_baixas_normalizado"]
) / 3

# Ordenar categorias da maior para a menor prioridade
df_categorias = df_categorias.sort_values(
    by="pontuacao_prioridade",
    ascending=False
)

print("Categorias com maior prioridade:")
print(
    df_categorias[
        [
            "product_category_name",
            "quantidade_itens",
            "percentual_atrasos",
            "percentual_avaliacoes_baixas",
            "pontuacao_prioridade"
        ]
    ].head(10)
)
```

```python


def preparar_dados(df_orders, df_order_items, df_products, df_order_reviews):

    pedidos = df_orders.copy()
    itens = df_order_items.copy()
    produtos = df_products.copy()
    avaliacoes = df_order_reviews.copy()

    # 1. Tratar categorias ausentes e desconhecidas
    produtos["product_category_name"] = (
        produtos["product_category_name"].fillna("sem_categoria")
    )

    # Garantir que "sem_categoria" faça parte das categorias de referência
    categorias_pipeline = list(categorias_referencia)

    if "sem_categoria" not in categorias_pipeline:
        categorias_pipeline.append("sem_categoria")

    produtos.loc[
        ~produtos["product_category_name"].isin(categorias_pipeline),
        "product_category_name"
    ] = "sem_categoria"

    # 2. Converter datas
    colunas_datas = [
        "order_purchase_timestamp",
        "order_approved_at",
        "order_delivered_carrier_date",
        "order_delivered_customer_date",
        "order_estimated_delivery_date"
    ]

    for coluna in colunas_datas:
        if coluna in pedidos.columns:
            pedidos[coluna] = pd.to_datetime(
                pedidos[coluna], errors="coerce"
            )

    if "shipping_limit_date" in itens.columns:
        itens["shipping_limit_date"] = pd.to_datetime(
            itens["shipping_limit_date"], errors="coerce"
        )

    for coluna in ["review_creation_date", "review_answer_timestamp"]:
        if coluna in avaliacoes.columns:
            avaliacoes[coluna] = pd.to_datetime(
                avaliacoes[coluna], errors="coerce"
            )

    # Garantir que as variáveis numéricas estejam no formato correto
    itens["price"] = pd.to_numeric(itens["price"], errors="coerce")
    itens["freight_value"] = pd.to_numeric(
        itens["freight_value"], errors="coerce"
    )
    avaliacoes["review_score"] = pd.to_numeric(
        avaliacoes["review_score"], errors="coerce"
    )

    # 3. Agrupar avaliações por pedido
    reviews = avaliacoes.groupby("order_id").agg(
        review_score=("review_score", "mean")
    ).reset_index()

    # 4. Unir os dados
    df_final = itens.merge(
        produtos[["product_id", "product_category_name"]],
        on="product_id",
        how="left"
    )

    df_final = df_final.merge(
        pedidos[
            [
                "order_id",
                "order_status",
                "order_delivered_customer_date",
                "order_estimated_delivery_date"
            ]
        ],
        on="order_id",
        how="left"
    )

    df_final = df_final.merge(
        reviews,
        on="order_id",
        how="left"
    )

    # Tratar categorias que continuaram ausentes após as junções
    df_final["product_category_name"] = (
        df_final["product_category_name"].fillna("sem_categoria")
    )

    # 5. Criar indicador de pedidos atrasados
    df_final["pedido_atrasado"] = pd.Series(
        pd.NA, index=df_final.index, dtype="Int64"
    )

    datas_validas = (
        df_final["order_status"].eq("delivered") &
        df_final["order_delivered_customer_date"].notna() &
        df_final["order_estimated_delivery_date"].notna()
    )

    df_final.loc[datas_validas, "pedido_atrasado"] = (
        df_final.loc[
            datas_validas, "order_delivered_customer_date"
        ] >
        df_final.loc[
            datas_validas, "order_estimated_delivery_date"
        ]
    ).astype(int)

    # 6. Criar indicador de avaliações baixas
    df_final["avaliacao_baixa"] = pd.Series(
        pd.NA, index=df_final.index, dtype="Int64"
    )

    avaliacoes_validas = df_final["review_score"].notna()

    df_final.loc[avaliacoes_validas, "avaliacao_baixa"] = (
        df_final.loc[
            avaliacoes_validas, "review_score"
        ] <= 2
    ).astype(int)

    # 7. Criar variável proporcional do frete
    df_final["frete_percentual_preco"] = (
        df_final["freight_value"]
        .div(df_final["price"])
        .where(df_final["price"] > 0)
        * 100
    )

    # 8. Criar valor total do item
    df_final["valor_total_item"] = (
        df_final["price"] + df_final["freight_value"]
    )

    # 9. Codificar categorias com One-Hot Encoding
    df_final["product_category_name"] = pd.Categorical(
        df_final["product_category_name"],
        categories=categorias_pipeline
    )

    dummies = pd.get_dummies(
        df_final["product_category_name"],
        prefix="categoria",
        drop_first=True,
        dtype=int
    )

    colunas_dummies = [
        "categoria_" + categoria
        for categoria in categorias_pipeline[1:]
    ]

    dummies = dummies.reindex(
        columns=colunas_dummies,
        fill_value=0
    )

    df_final = pd.concat([df_final, dummies], axis=1)

    # 10. Aplicar parâmetros fixos de Min-Max e Z-Score
    for coluna in colunas_numericas:

        minimo = parametros_escala[coluna]["minimo"]
        maximo = parametros_escala[coluna]["maximo"]
        media = parametros_escala[coluna]["media"]
        desvio = parametros_escala[coluna]["desvio"]

        # Min-Max
        if maximo != minimo:
            df_final[coluna + "_minmax"] = (
                (df_final[coluna] - minimo) / (maximo - minimo)
            )
        else:
            df_final[coluna + "_minmax"] = 0

        # Z-Score
        if desvio != 0:
            df_final[coluna + "_zscore"] = (
                (df_final[coluna] - media) / desvio
            )
        else:
            df_final[coluna + "_zscore"] = 0

    return df_final
```

```python

# Indicadores agrupados por categoria

df_categorias_final = df_teste_final.groupby(
    "product_category_name",
    observed=True
).agg(
    quantidade_itens=("order_id", "count"),
    percentual_frete_medio=("frete_percentual_preco", "mean"),
    percentual_atrasos=("pedido_atrasado", "mean"),
    percentual_avaliacoes_baixas=("avaliacao_baixa", "mean")
).reset_index()

# Converter os indicadores em porcentagem
df_categorias_final["percentual_atrasos"] *= 100

df_categorias_final["percentual_avaliacoes_baixas"] *= 100

# Ordenar por volume de itens
df_categorias_final = df_categorias_final.sort_values(
    by="quantidade_itens",
    ascending=False
).reset_index(drop=True)

print("INDICADORES POR CATEGORIA:")
display(df_categorias_final.head(10))

print("\nTotal de categorias:", len(df_categorias_final))

print("\nValores ausentes:")
print(
    df_categorias_final.isnull().sum()[
        df_categorias_final.isnull().sum() > 0
    ]
)
```
